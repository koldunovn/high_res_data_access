```python
import matplotlib.pyplot as plt
```

```python
# Data of the run with 64GB allocated memory, scaling with 2, 4, 8, and 16 processes
# The execution times are in seconds.
# The data is for the variable '2t' from the model 'IFS-FESOM'
# The experiment is 'story-2017-historical-HPC' with a regrid of 'r100'
# The source is 'hourly-hpz9-atm2d' and the frequency is 'monthly'

nprocs_64 = (2,4,8,16,32)
execution_times_64 = (1346.10,758.33,489.50,370.63,291.30)
nprocs_32 = (2,4,8,16)
execution_times_32 = (1521.34,1009.19,723.81,466.03)
nprocs_128 = (2,4,8,16,32)
execution_times_128 = (1215.15,636.47,471.40,443.22,295.45)
```

```python
# plotting the data

plt.plot(nprocs_64, execution_times_64, marker='o', label='64GB Memory Allocation')
plt.plot(nprocs_32, execution_times_32, marker='o', label='32GB Memory Allocation')
plt.plot(nprocs_128, execution_times_128, marker='o', label='128GB Memory Allocation')
plt.xlabel('Number of Processes')
plt.ylabel('Execution Time (s)')
plt.title('Execution Time vs Number of Processes (varying memory allocation)')
plt.legend()
plt.xscale('log')
plt.yscale('log')
```
