```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Data of the run with 64GB allocated memory, scaling with 2, 4, 8, and 16 processes
# The execution times are in seconds.
# The data is for the variable '2t' from the model 'IFS-FESOM'
# The experiment is 'story-2017-historical-HPC' with a regrid of 'r100'
# The source is 'hourly-hpz9-atm2d' and the frequency is 'monthly'

nprocs_64 = (2,4,8,16,32)
execution_times_64 = (1346.10,758.33,489.50,370.63,291.30)
nprocs_32 = (2,4,8,16)
execution_times_32 = (1521.34,1009.19,723.81,466.03)
nprocs_128 = (2,4,8,16,32,64)
execution_times_128 = (1215.15,636.47,471.40,443.22,295.45,242.28)
nprocs_16 = (2,4,8)
execution_times_16 = (1283.24,776.35,710.76)
nprocs_8 = (2,4)
execution_times_8 = (1197.56,776.08)
```

```python
# plotting the data

plt.plot(nprocs_64, execution_times_64, marker='o', label='64GB Memory Allocation')
plt.plot(nprocs_32, execution_times_32, marker='o', label='32GB Memory Allocation')
plt.plot(nprocs_128, execution_times_128, marker='o', label='128GB Memory Allocation')
plt.plot(nprocs_16, execution_times_16, marker='o', label='16GB Memory Allocation')
plt.plot(nprocs_8, execution_times_8, marker='o', label='8GB Memory Allocation')
plt.xlabel('Number of Processes')
plt.ylabel('Execution Time (s)')
plt.title('Execution Time vs Number of Processes (varying memory allocation)')
plt.legend()
#plt.xscale('log')
#plt.yscale('log')
```

```python
plt.plot(nprocs_128, np.divide(execution_times_128[0], execution_times_128), marker='o', label='128GB Memory Allocation (normalized)')
plt.plot(nprocs_64, np.divide(execution_times_64[0], execution_times_64), marker='o', label='64GB Memory Allocation (normalized)')
plt.plot(nprocs_32, np.divide(execution_times_32[0], execution_times_32), marker='o', label='32GB Memory Allocation (normalized)')
plt.plot(nprocs_16, np.divide(execution_times_16[0], execution_times_16), marker='o', label='16GB Memory Allocation (normalized)')
plt.plot(nprocs_8, np.divide(execution_times_8[0], execution_times_8), marker='o', label='8GB Memory Allocation (normalized)')
plt.xlabel('Number of Processes')
plt.ylabel('Execution Time (s)')
plt.title('Execution Time vs Number of Processes (varying memory allocation)')

plt.plot(nprocs_64, np.divide(nprocs_64,2), linestyle='--', color='gray', label='Perfect Scaling (64GB)')

# Add perfect scaling line
# perfect_scaling = [execution_times_8[0] / n for n in nprocs_64]
# plt.plot(nprocs_64, perfect_scaling, linestyle='--', color='gray', label='Perfect Scaling')

#plt.xscale('log')
#plt.yscale('log')

plt.legend()
```
