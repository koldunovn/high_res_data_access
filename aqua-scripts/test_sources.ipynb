```python
from aqua import Reader
from aqua.graphics import plot_single_map
import time
```

```python
catalog = 'data-access'
model='IFS-FESOM'
exp='story-2017-historical-stac'
source='hourly-hpz9-atm2d'
```

```python
# time the cell
start_time = time.time()    

# create a reader object
reader = Reader(catalog=catalog, model=model, exp=exp, source=source, loglevel='WARNING')

# read the file
data = reader.retrieve()
plot_single_map(data.isel(time=0)['2t'])

# print the time taken to read the file
print("--- %s seconds ---" % (time.time() - start_time))
```
